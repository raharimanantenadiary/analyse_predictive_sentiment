```python
# ========================================
# ÉTAPE 3 : NETTOYAGE ET PREPROCESSING NLP
# Projet : Analyse de Sentiment Twitter
# ========================================

import pandas as pd
import numpy as np
import re
import string
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

print("📦 Chargement des ressources NLTK...")
nltk.download('stopwords', quiet=True)
nltk.download('punkt', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)
print("✅ Ressources chargées !\n")
```

```python
# ========================================
# 1. CHARGEMENT DES DONNÉES
# ========================================

print("📂 Chargement des données...")
donnees = pd.read_csv("tweets.csv")  # ← Change le nom de ton fichier

# IMPORTANT : Modifie ces noms selon TES colonnes
# Exemple : si ta colonne s'appelle "tweet_text", mets "tweet_text"
COLONNE_TEXTE = "text"  # ← Nom de la colonne contenant les tweets
COLONNE_SENTIMENT = "sentiment"  # ← Nom de la colonne contenant les sentiments

print(f"✅ {len(donnees)} tweets chargés")
print(f"📊 Colonnes utilisées : '{COLONNE_TEXTE}' et '{COLONNE_SENTIMENT}'")

# ========================================
# 2. CRÉATION D'UNE COPIE DE TRAVAIL
# ========================================

donnees_propres = donnees.copy()
print(f"\n🔧 Copie de travail créée : {len(donnees_propres)} lignes")
```

```python
# ========================================
# 3. SUPPRESSION DES VALEURS MANQUANTES
# ========================================

print("\n" + "=" * 50)
print("🧹 NETTOYAGE DES VALEURS MANQUANTES")
print("=" * 50)

avant = len(donnees_propres)
donnees_propres = donnees_propres.dropna(subset=[COLONNE_TEXTE, COLONNE_SENTIMENT])
apres = len(donnees_propres)

print(f"Avant : {avant} tweets")
print(f"Après : {apres} tweets")
print(f"❌ Supprimés : {avant - apres} tweets")
```

```python
# ========================================
# 4. FONCTION DE NETTOYAGE DE TEXTE
# ========================================

def nettoyer_texte(texte):
    """
    Nettoie un tweet en supprimant les éléments inutiles
    
    Arguments:
        texte (str) : Le texte à nettoyer
    
    Retourne:
        str : Le texte nettoyé
    """
    
    # Convertir en minuscules
    texte = texte.lower()
    
    # Supprimer les URLs (http, https, www)
    texte = re.sub(r'http\S+|www\S+|https\S+', '', texte, flags=re.MULTILINE)
    
    # Supprimer les mentions (@username)
    texte = re.sub(r'@\w+', '', texte)
    
    # Supprimer les hashtags (#hashtag) - optionnel
    # texte = re.sub(r'#\w+', '', texte)
    
    # Supprimer les hashtags mais garder le mot
    texte = re.sub(r'#', '', texte)
    
    # Supprimer les chiffres
    texte = re.sub(r'\d+', '', texte)
    
    # Supprimer la ponctuation
    texte = texte.translate(str.maketrans('', '', string.punctuation))
    
    # Supprimer les espaces multiples
    texte = re.sub(r'\s+', ' ', texte)
    
    # Supprimer les espaces au début et à la fin
    texte = texte.strip()
    
    return texte
```

```python
# ========================================
# 5. APPLICATION DU NETTOYAGE
# ========================================

print("\n" + "=" * 50)
print("🧼 NETTOYAGE DES TWEETS")
print("=" * 50)

# Exemple AVANT nettoyage
print("\n📝 EXEMPLE AVANT NETTOYAGE :")
tweet_exemple = donnees_propres[COLONNE_TEXTE].iloc[0]
print(f"'{tweet_exemple}'")

# Appliquer le nettoyage à tous les tweets
donnees_propres['texte_nettoye'] = donnees_propres[COLONNE_TEXTE].apply(nettoyer_texte)

# Exemple APRÈS nettoyage
print("\n✨ EXEMPLE APRÈS NETTOYAGE :")
print(f"'{donnees_propres['texte_nettoye'].iloc[0]}'")
```

```python
# ========================================
# 6. SUPPRESSION DES STOPWORDS
# ========================================

print("\n" + "=" * 50)
print("🚫 SUPPRESSION DES STOPWORDS (mots vides)")
print("=" * 50)

# Charger les stopwords anglais
mots_vides = set(stopwords.words('english'))

# Ajouter des stopwords personnalisés si nécessaire
mots_vides_personnalises = {'rt', 'via', 'amp'}  # Mots spécifiques à Twitter
mots_vides.update(mots_vides_personnalises)

print(f"📋 Nombre de stopwords : {len(mots_vides)}")
print(f"📝 Exemples : {list(mots_vides)[:10]}")

def supprimer_stopwords(texte):
    """
    Supprime les stopwords d'un texte
    
    Arguments:
        texte (str) : Le texte à traiter
    
    Retourne:
        str : Le texte sans stopwords
    """
    # Tokenisation (découper en mots)
    mots = texte.split()
    
    # Garder seulement les mots qui ne sont pas des stopwords
    mots_filtres = [mot for mot in mots if mot not in mots_vides]
    
    # Reconstituer le texte
    return ' '.join(mots_filtres)

# Appliquer la suppression des stopwords
donnees_propres['texte_sans_stopwords'] = donnees_propres['texte_nettoye'].apply(supprimer_stopwords)

# Exemple
print("\n📝 EXEMPLE AVANT suppression stopwords :")
print(f"'{donnees_propres['texte_nettoye'].iloc[0]}'")
print("\n✨ EXEMPLE APRÈS suppression stopwords :")
print(f"'{donnees_propres['texte_sans_stopwords'].iloc[0]}'")
```

```python
# ========================================
# 7. LEMMATISATION
# ========================================

print("\n" + "=" * 50)
print("🔤 LEMMATISATION (réduction à la racine)")
print("=" * 50)

lemmatiseur = WordNetLemmatizer()

def lemmatiser_texte(texte):
    """
    Lemmatise les mots (ramène à leur forme de base)
    Exemple : "running" → "run", "better" → "good"
    
    Arguments:
        texte (str) : Le texte à lemmatiser
    
    Retourne:
        str : Le texte lemmatisé
    """
    mots = texte.split()
    mots_lemmatises = [lemmatiseur.lemmatize(mot) for mot in mots]
    return ' '.join(mots_lemmatises)

# Appliquer la lemmatisation
print("⏳ Lemmatisation en cours... (peut prendre quelques secondes)")
donnees_propres['texte_final'] = donnees_propres['texte_sans_stopwords'].apply(lemmatiser_texte)
print("✅ Lemmatisation terminée !")

# Exemple
print("\n📝 EXEMPLE AVANT lemmatisation :")
print(f"'{donnees_propres['texte_sans_stopwords'].iloc[0]}'")
print("\n✨ EXEMPLE APRÈS lemmatisation :")
print(f"'{donnees_propres['texte_final'].iloc[0]}'")
```

```python
# ========================================
# 8. SUPPRESSION DES TWEETS VIDES
# ========================================

print("\n" + "=" * 50)
print("🗑️ SUPPRESSION DES TWEETS VIDES")
print("=" * 50)

avant = len(donnees_propres)

# Supprimer les tweets vides après nettoyage
donnees_propres = donnees_propres[donnees_propres['texte_final'].str.strip() != '']

apres = len(donnees_propres)

print(f"Avant : {avant} tweets")
print(f"Après : {apres} tweets")
print(f"❌ Supprimés : {avant - apres} tweets vides")
```

```python
# ========================================
# 2. CRÉATION DU VECTORISEUR TF-IDF
# ========================================

print("\n" + "=" * 50)
print("🔧 CRÉATION DU VECTORISEUR TF-IDF")
print("=" * 50)

# Créer le vectoriseur avec des paramètres optimaux
vectoriseur_tfidf = TfidfVectorizer(
    max_features=5000,        # Garder les 5000 mots les plus importants
    min_df=2,                 # Un mot doit apparaître dans au moins 2 documents
    max_df=0.8,               # Ignorer les mots qui apparaissent dans plus de 80% des documents
    ngram_range=(1, 2),       # Utiliser unigrammes (1 mot) et bigrammes (2 mots)
    sublinear_tf=True         # Échelle logarithmique pour TF
)

print("✅ Vectoriseur TF-IDF créé avec les paramètres :")
print(f"   - Nombre maximum de features : 5000")
print(f"   - Fréquence minimum : 2 documents")
print(f"   - Fréquence maximum : 80% des documents")
print(f"   - N-grams : unigrammes et bigrammes")
```

```python
# 3. TRANSFORMATION DES TEXTES EN VECTEURS
# ========================================

print("\n" + "=" * 50)
print("🔄 TRANSFORMATION DES TEXTES EN VECTEURS")
print("=" * 50)

print("⏳ Vectorisation en cours...")

# Transformer les textes en vecteurs TF-IDF
X_vectorise = vectoriseur_tfidf.fit_transform(X)

print("✅ Vectorisation terminée !")
print(f"📊 Forme de la matrice : {X_vectorise.shape}")
print(f"   - {X_vectorise.shape[0]} tweets")
print(f"   - {X_vectorise.shape[1]} features (mots)")
```

```python
# ========================================
# 4. AFFICHAGE DES MOTS IMPORTANTS
# ========================================

print("\n" + "=" * 50)
print("📝 TOP 20 MOTS LES PLUS IMPORTANTS")
print("=" * 50)

# Récupérer les noms des features (mots)
noms_features = vectoriseur_tfidf.get_feature_names_out()
print(f"Vocabulaire total : {len(noms_features)} mots/bigrammes")

# Calculer l'importance moyenne de chaque mot
importance_moyenne = X_vectorise.mean(axis=0).A1
mots_importantes = sorted(
    zip(noms_features, importance_moyenne), 
    key=lambda x: x[1], 
    reverse=True
)

print("\nTop 20 mots/bigrammes par importance TF-IDF :")
for i, (mot, score) in enumerate(mots_importantes[:20], 1):
    print(f"{i:2d}. {mot:20s} → Score: {score:.6f}")
```

```python
# 5. DIVISION DES DONNÉES (TRAIN/TEST)
# ========================================

print("\n" + "=" * 50)
print("✂️ DIVISION DES DONNÉES (TRAIN/TEST)")
print("=" * 50)

# Diviser en 80% entraînement et 20% test
X_train, X_test, y_train, y_test = train_test_split(
    X_vectorise, 
    y, 
    test_size=0.2,           # 20% pour le test
    random_state=42,         # Pour la reproductibilité
    stratify=y               # Garde la même proportion de classes
)

print("✅ Division effectuée !")
print(f"\n📊 Ensemble d'ENTRAÎNEMENT :")
print(f"   - Tweets : {X_train.shape[0]}")
print(f"   - Features : {X_train.shape[1]}")

print(f"\n📊 Ensemble de TEST :")
print(f"   - Tweets : {X_test.shape[0]}")
print(f"   - Features : {X_test.shape[1]}")
```

```python
# ========================================
# 6. DISTRIBUTION DES CLASSES
# ========================================

print("\n" + "=" * 50)
print("📊 DISTRIBUTION DES CLASSES")
print("=" * 50)

print("\n🎯 Ensemble d'ENTRAÎNEMENT :")
distribution_train = pd.Series(y_train).value_counts()
for classe, nombre in distribution_train.items():
    pourcentage = (nombre / len(y_train)) * 100
    print(f"   {classe} : {nombre} tweets ({pourcentage:.2f}%)")

print("\n🎯 Ensemble de TEST :")
distribution_test = pd.Series(y_test).value_counts()
for classe, nombre in distribution_test.items():
    pourcentage = (nombre / len(y_test)) * 100
    print(f"   {classe} : {nombre} tweets ({pourcentage:.2f}%)")
```

```python
# ========================================
# 7. EXEMPLE DE VECTORISATION
# ========================================

print("\n" + "=" * 50)
print("🔬 EXEMPLE DE VECTORISATION")
print("=" * 50)

# Prendre un exemple
indice_exemple = 0
texte_exemple = X.iloc[indice_exemple]
sentiment_exemple = y.iloc[indice_exemple]

print(f"\n📝 Texte original :")
print(f"'{texte_exemple}'")
print(f"Sentiment : {sentiment_exemple}")

# Transformer ce texte
vecteur_exemple = vectoriseur_tfidf.transform([texte_exemple])

# Trouver les mots avec les scores TF-IDF les plus élevés
vecteur_dense = vecteur_exemple.toarray()[0]
indices_non_nuls = vecteur_dense.nonzero()[0]

print(f"\n🔢 Représentation vectorielle :")
print(f"Taille du vecteur : {vecteur_exemple.shape[1]}")
print(f"Valeurs non-nulles : {len(indices_non_nuls)}")

print("\n📈 Top 5 mots avec les scores TF-IDF les plus élevés :")
mots_scores = [(noms_features[i], vecteur_dense[i]) for i in indices_non_nuls]
mots_scores_tries = sorted(mots_scores, key=lambda x: x[1], reverse=True)

for i, (mot, score) in enumerate(mots_scores_tries[:5], 1):
    print(f"{i}. {mot:20s} → Score: {score:.6f}")
```

```python
# ========================================
# ÉTAPE 5 : ENTRAÎNEMENT DU MODÈLE
# Projet : Analyse de Sentiment Twitter
# ========================================

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
```

```python
# ========================================
# 1. CRÉATION DU MODÈLE
# ========================================

print("\n" + "=" * 50)
print("🤖 CRÉATION DU MODÈLE")
print("=" * 50)

# Créer le modèle de Régression Logistique
modele = LogisticRegression(
    max_iter=1000,           # Nombre maximum d'itérations
    random_state=42,         # Pour la reproductibilité
    solver='lbfgs',          # Algorithme d'optimisation
    multi_class='multinomial', # Pour la classification multi-classes
    n_jobs=-1                # Utiliser tous les CPU disponibles
)

print("✅ Modèle créé : Régression Logistique")
print("   - Type : Classification multi-classes")
print("   - Solver : lbfgs")
print("   - Max iterations : 1000")
```

```python
# ========================================
# 2. ENTRAÎNEMENT DU MODÈLE
# ========================================

print("\n" + "=" * 50)
print("🏋️ ENTRAÎNEMENT DU MODÈLE")
print("=" * 50)

print(f"⏳ Entraînement en cours sur {X_train.shape[0]} tweets...")
print(f"   Features utilisées : {X_train.shape[1]}")

# Entraîner le modèle
modele.fit(X_train, y_train)

print("✅ Entraînement terminé !")
```

```python
# ========================================
# 3. PRÉDICTIONS SUR L'ENSEMBLE DE TEST
# ========================================

print("\n" + "=" * 50)
print("🔮 PRÉDICTIONS SUR L'ENSEMBLE DE TEST")
print("=" * 50)

# Faire des prédictions
y_pred = modele.predict(X_test)

print(f"✅ {len(y_pred)} prédictions effectuées")

# Afficher quelques exemples
print("\n📝 Exemples de prédictions (10 premiers) :")
print(f"{'Réel':<15} | {'Prédit':<15} | {'Correct'}")
print("-" * 45)
for i in range(min(10, len(y_test))):
    reel = y_test.iloc[i]
    predit = y_pred[i]
    correct = "✅" if reel == predit else "❌"
    print(f"{reel:<15} | {predit:<15} | {correct}")
```

```python
# ========================================
# 4. CALCUL DES MÉTRIQUES DE PERFORMANCE
# ========================================

print("\n" + "=" * 50)
print("📊 MÉTRIQUES DE PERFORMANCE")
print("=" * 50)

# Calculer les métriques
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')
f1 = f1_score(y_test, y_pred, average='weighted')

print(f"\n🎯 ACCURACY (Exactitude)  : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"🎯 PRECISION (Précision)  : {precision:.4f} ({precision*100:.2f}%)")
print(f"🎯 RECALL (Rappel)        : {recall:.4f} ({recall*100:.2f}%)")
print(f"🎯 F1-SCORE               : {f1:.4f} ({f1*100:.2f}%)")
```

```python
# ========================================
# 5. RAPPORT DE CLASSIFICATION DÉTAILLÉ
# ========================================

print("\n" + "=" * 50)
print("📋 RAPPORT DE CLASSIFICATION DÉTAILLÉ")
print("=" * 50)

# Rapport détaillé par classe
rapport = classification_report(y_test, y_pred, target_names=modele.classes_)
print("\n" + rapport)
```

```python
# ========================================
# 6. MATRICE DE CONFUSION
# ========================================

print("\n" + "=" * 50)
print("🔢 MATRICE DE CONFUSION")
print("=" * 50)

# Calculer la matrice de confusion
matrice_confusion = confusion_matrix(y_test, y_pred)

print("\nMatrice de confusion :")
print(matrice_confusion)

# Créer un graphique de la matrice de confusion
plt.figure(figsize=(10, 8))
sns.heatmap(
    matrice_confusion, 
    annot=True,           # Afficher les valeurs
    fmt='d',              # Format entier
    cmap='Blues',         # Palette de couleurs
    xticklabels=modele.classes_,
    yticklabels=modele.classes_,
    cbar_kws={'label': 'Nombre de prédictions'}
)
plt.title('Matrice de Confusion', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Sentiment Prédit', fontsize=12, fontweight='bold')
plt.ylabel('Sentiment Réel', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig('matrice_confusion.png', dpi=300, bbox_inches='tight')
plt.show()

print("✅ Graphique sauvegardé : 'matrice_confusion.png'")
```

```python
# ========================================
# 7. ANALYSE DES ERREURS
# ========================================

print("\n" + "=" * 50)
print("🔍 ANALYSE DES ERREURS")
print("=" * 50)

# Trouver les prédictions incorrectes
indices_erreurs = y_test.reset_index(drop=True) != pd.Series(y_pred)
nombre_erreurs = indices_erreurs.sum()
nombre_correct = len(y_test) - nombre_erreurs

print(f"✅ Prédictions correctes : {nombre_correct} / {len(y_test)} ({(nombre_correct/len(y_test)*100):.2f}%)")
print(f"❌ Prédictions incorrectes : {nombre_erreurs} / {len(y_test)} ({(nombre_erreurs/len(y_test)*100):.2f}%)")

# Afficher quelques exemples d'erreurs
if nombre_erreurs > 0:
    print("\n📝 Exemples d'ERREURS (5 premiers) :")
    
    # ✅ ICI LA CORRECTION
    X_original_test = X.loc[y_test.index].reset_index(drop=True)
    y_test_reset = y_test.reset_index(drop=True)
    
    compteur_erreurs = 0
    for i in range(len(y_test)):
        if indices_erreurs.iloc[i] and compteur_erreurs < 5:
            print(f"\n--- Erreur {compteur_erreurs + 1} ---")
            print(f"Texte : {X_original_test.iloc[i][:80]}...")
            print(f"Réel : {y_test_reset.iloc[i]}")
            print(f"Prédit : {y_pred[i]}")
            compteur_erreurs += 1
```

```python
# ========================================
# 8. IMPORTANCE DES FEATURES
# ========================================

print("\n" + "=" * 50)
print("🔝 MOTS LES PLUS IMPORTANTS PAR SENTIMENT")
print("=" * 50)

# Récupérer les coefficients du modèle
coefficients = modele.coef_
noms_features = vectoriseur_tfidf.get_feature_names_out()

# Pour chaque classe (sentiment)
for i, classe in enumerate(modele.classes_):
    print(f"\n📌 Sentiment : {classe}")
    
    # Récupérer les coefficients pour cette classe
    coefs_classe = coefficients[i]
    
    # Trouver les 10 mots les plus importants (positifs)
    indices_top = coefs_classe.argsort()[-10:][::-1]
    
    print("   Top 10 mots caractéristiques :")
    for j, idx in enumerate(indices_top, 1):
        mot = noms_features[idx]
        score = coefs_classe[idx]
        print(f"   {j:2d}. {mot:20s} → Coefficient: {score:.4f}")
```

```python
# ========================================
# 11. TEST SUR DE NOUVEAUX TEXTES
# ========================================

print("\n" + "=" * 50)
print("🧪 TEST SUR DE NOUVEAUX TWEETS")
print("=" * 50)

# Fonction pour prédire le sentiment d'un nouveau tweet
def predire_sentiment(texte_nouveau):
    """
    Prédit le sentiment d'un nouveau tweet
    
    Arguments:
        texte_nouveau (str) : Le tweet à analyser
    
    Retourne:
        str : Le sentiment prédit
    """
    # Nettoyer le texte (utilise les fonctions de l'étape 3)
    texte_nettoye = nettoyer_texte(texte_nouveau)
    texte_sans_stopwords = supprimer_stopwords(texte_nettoye)
    texte_final = lemmatiser_texte(texte_sans_stopwords)
    
    # Vectoriser
    texte_vectorise = vectoriseur_tfidf.transform([texte_final])
    
    # Prédire
    prediction = modele.predict(texte_vectorise)[0]
    probabilites = modele.predict_proba(texte_vectorise)[0]
    
    return prediction, probabilites

# Exemples de tests
tweets_test = [
    "I love this product! It's amazing and works perfectly!",
    "This is the worst experience ever. Totally disappointed.",
    "It's okay, nothing special but not bad either."
]

print("\n📝 Exemples de prédictions :")
for i, tweet in enumerate(tweets_test, 1):
    sentiment, probas = predire_sentiment(tweet)
    print(f"\n{i}. Tweet : {tweet}")
    print(f"   Sentiment prédit : {sentiment}")
    print(f"   Probabilités : {dict(zip(modele.classes_, probas))}")
```
